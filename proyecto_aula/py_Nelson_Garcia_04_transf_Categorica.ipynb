Se utiliza la técnica Label Encoding porque genera un orden implícito en la importancia del tipo de combustible.


```python
# Uso de Librerías
import pandas as pd
from sklearn.preprocessing import LabelEncoder
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Configuración de estilo
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (8,4)

# Estilo general
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))

# Crear dataset
df = pd.read_csv("../datos/dataset_depurado_sin_duplicados.csv", sep=";")
```

```python
df_encoded = df.copy()
label_encoders = {}

for col in df_encoded.select_dtypes(include="object"):  # solo categóricas
    le = LabelEncoder()
    df_encoded[col] = le.fit_transform(df_encoded[col])
    label_encoders[col] = le

print("\nDatos después de aplicar Label Encoding:")
print(label_encoders)
```

```python
# Revertir los valores codificados a sus etiquetas originales
df_encoded = pd.DataFrame(df_encoded, columns=label_encoders.keys())
df_decoded = df_encoded.copy()
for col in df_decoded.columns:
    le = label_encoders[col]
    df_decoded[col] = le.inverse_transform(df_decoded[col])

print("\nDatos revertidos a sus etiquetas originales:")
print(df_decoded)
```

### <span style="color:Black;"><strong>Interpretación</strong></span>

La variable Fuel (combustible) se incluye en el modelo debido a su importancia en la determinación del precio y desempeño del vehículo. El tipo de combustible influye en el consumo, los costos de mantenimiento y la valoración del mercado, lo que la convierte en un predictor relevante. Dado que es una variable categórica, fue necesario aplicarle una transformación mediante codificación (label Encoding) para permitir su uso en algoritmos predictivos y análisis multivariado que requieren valores numéricos. Esta transformación garantiza que el modelo procese la variable correctamente sin introducir jerarquías artificiales ni sesgos, y mejora la calidad de las predicciones y la estabilidad del análisis.
